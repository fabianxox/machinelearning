```python
import numpy as np

import numpy as np

# Features:
# [Acting Skill, Charisma, Fitness, Screen Presence]

x = np.array([
    [3, 4, 5, 4],   # struggling actor
    [4, 5, 6, 5],
    [6, 6, 7, 6],
    [7, 8, 7, 8],
    [2, 3, 4, 3],   # extra in background
    [8, 9, 8, 9],   # rising star
    [9, 9, 9, 9],   # superstar
    [5, 6, 5, 6],
    [10, 10, 9, 10] # Brad Pitt style profile
])

# Target:
# 1 = Blockbuster Star
# 0 = Not a star

y = np.array([
    [0],
    [0],
    [0],
    [1],
    [0],
    [1],
    [1],
    [0],
    [1]
])
```

```python
input_layer= 4
hidden_layer= 5
output_layer=1

np.random.seed(42)

global w1, b1, w2, b2

w1= np.random.rand(input_layer, hidden_layer)
b1= np.zeros((1, hidden_layer))

w2= np.random.rand(hidden_layer, output_layer)
b2= np.zeros((1, output_layer))
```

```python
#normalizing

x= x/10
```

```python
# sigmmoid actvation

def sigmoid(x):
  return 1/ (1+ np.exp(-x))
```

```python
def sigmoid_derivative(x):
    return x * (1 - x)
```

```python
def forward_prop(x):
    global z1, a1, z2, a2

    z1 = np.dot(x, w1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, w2) + b2
    a2 = sigmoid(z2)

    return a2
```

```python
def loss(y, y_pred):
    return np.mean((y_pred - y) ** 2)
```

```python
def back_prop(x, y):
    global w1, b1, w2, b2

    learning_rate = 0.1

    output_error = a2 - y
    output_gradient = output_error * sigmoid_derivative(a2)

    hidden_error = np.dot(output_gradient, w2.T)
    hidden_gradient = hidden_error * sigmoid_derivative(a1)

    w2 -= np.dot(a1.T, output_gradient) * learning_rate
    b2 -= np.sum(output_gradient, axis=0, keepdims=True) * learning_rate

    w1 -= np.dot(x.T, hidden_gradient) * learning_rate
    b1 -= np.sum(hidden_gradient, axis=0, keepdims=True) * learning_rate
```

```python
epochs= 3000

for i in range(3000):
  predictions= forward_prop(x)
  back_prop(x, y)

  if i%300 == 0:
    print(f"epoch: {i}, cost: {loss(y, predictions)}")
```

```python
test= np.zeros(4)
questions= ["Acting Skill", "Charisma", "Fitness", "Screen Presence"]
for i in range(4):
    test[i]= (int(input(questions[i])))
test= test/10
predict= forward_prop(test)
if(predict> 0.5):
  print("blockbuster hero")
else:
  print("avg one")
```
