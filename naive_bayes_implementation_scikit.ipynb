```python
import pandas as pd
from sklearn.naive_bayes import BernoulliNB
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import accuracy_score
```

```python
data = pd.DataFrame({
    "Age":[22,25,24,45,50,47],
    "Salary":[30000,35000,32000,80000,90000,85000],
    "Class":["No","No","No","Yes","Yes","Yes"]
})
```

```python
X= data[["Age", "Salary"]]
y= data["Class"]
```

```python
model= GaussianNB()
# nothing big happens here, the imported gaussian model has been pointed by a variable
```

```python
model.fit(X, y)
# training the model with the dataset created above
```

```python
new_data = pd.DataFrame({
    "Age": [5],
    "Salary": [540500]
})
prediction= model.predict(new_data)
print(prediction)
```

```python
dataset_prediction= model.predict(X)

accuracy= accuracy_score(y, dataset_prediction)
print(accuracy)
```

```python
model_2.fit(X_2, y_2)
```

```python
prediction_2 = model_2.predict([[0, 0,1, 1]])
prediction_2_p = model_2.predict_proba([[0, 0,1, 1]])
print(prediction_2)
```
