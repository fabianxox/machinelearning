```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x= np.array([
    120,
    130,
    125,
    128,
    135,
    140,
    150,
    132,
    138
])
```

```python
#it is undupervised, so no label y
#compute mean and variance

mu= np.mean(x)
variance= np.var(x)
print(mu, variance)
```

```python
# now let us define the gaussian prob function, it says how normal is this number?
# if very low, it is anamoly

def gaussian_prob(x, mu, var):
  coeff= 1/ np.sqrt(2*np.pi*var)
  exp= np.exp(-((x-mu)**2)/(2*var))
  print(coeff*exp)
  return coeff * exp
```

```python
test_data = np.array([
    129,
    131,
    5000,
    930,
    127,
    3000
])
```

```python
epsilon= 0.00001
```

```python
for i in test_data:
  p= gaussian_prob(i, mu, variance)
  if p<epsilon:
    print("anamoly")
  else:
    print("normal")
```
