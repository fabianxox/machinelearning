```python

!pip -q install pandas matplotlib seaborn
!apt-get -qq install -y r-base
!Rscript -e 'install.packages("ggplot2", repos="https://cloud.r-project.org", quiet=TRUE)'

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import zipfile, glob, subprocess
from google.colab import files
from IPython.display import Image, display, Markdown

# Upload dataset
uploaded = files.upload()
with zipfile.ZipFile(next(iter(uploaded))) as z:
    z.extractall("data")

df = pd.read_csv(glob.glob(
    "data/**/firearm_homicides_country_year.csv",
    recursive=True
)[0])

sns.set_theme(style="whitegrid")

# ==============================
# 1. MATPLOTLIB
# ==============================

display(Markdown("# 1. MATPLOTLIB"))

trend = df.groupby("year").homicide_rate.mean()

plt.figure(figsize=(9, 5))
plt.plot(trend.index, trend.values, marker="o")
plt.title("Homicide Rate Over Time")
plt.xlabel("Year")
plt.ylabel("Average Homicide Rate")
plt.tight_layout()
plt.show()

# ==============================
# 2. SEABORN
# ==============================

display(Markdown("# 2. SEABORN"))

plt.figure(figsize=(9, 5))
sns.barplot(data=df, x="region", y="homicide_rate")
plt.title("Homicide Rate by Region")
plt.xlabel("Region")
plt.ylabel("Homicide Rate")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# ==============================
# 3. GGPLOT2
# ==============================

display(Markdown("# 3. GGPLOT2"))

df.to_csv("data.csv", index=False)

r_code = '''
library(ggplot2)
df <- read.csv("data.csv")

p <- ggplot(df, aes(year, homicide_rate)) +
  stat_summary(fun=mean, geom="line", color="blue") +
  stat_summary(fun=mean, geom="point") +
  labs(title="Homicide Rate Over Time",
       x="Year", y="Average Homicide Rate") +
  theme_minimal()

ggsave("plot.png", p, width=9, height=5)
'''

with open("plot.R", "w") as f:
    f.write(r_code)

subprocess.run(["Rscript", "plot.R"], check=True)
display(Image("plot.png"))

# ==============================
# END
# ==============================

display(Markdown("# PROJECT COMPLETED"))
```
