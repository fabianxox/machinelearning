```python
import pandas as pd
import math
```

```python
data = pd.DataFrame({
    "Age":[22,25,24,45,50,47],
    "Salary":[30000,35000,32000,80000,90000,85000],
    "Class":["No","No","No","Yes","Yes","Yes"]
})

print(data)
```

```python
def get_stats(data):
  stats={}
  classes = data["Class"].unique()
  prior_prob= {}
  for cls in classes:
    class_data= data[data["Class"]==cls]
    prior_prob[cls]= class_data.shape[0]/len(data)
    for feature in data.columns:
      if feature == "Class":
        continue
      mean= class_data[feature].mean()
      variance=0
      for value in class_data[feature]:
        variance+= (value-mean)**2
      variance/=len(class_data)
      variance = max(variance, 1e-9)
      if feature not in stats:
        stats[feature]= {}
      stats[feature][cls]={
          "mean": mean,
          "variance": variance
        }
  return stats, prior_prob
```

```python
stats, prior_probability= get_stats(data)
print(stats)
print(prior_probability)
```

```python
df= pd.DataFrame(stats)
print(df)
```

```python
def calculate_probability(x, mean, variance):
  prb = -((x - mean)**2 / (2 * variance))
  prb = math.exp(prb)
  return (1 / math.sqrt(2 * math.pi * variance)) * prb
```

```python
new_data = {
    "Age": 46,
    "Salary": 85000
}
```

```python
def predict(new_data, stats, calculate_probability, prior_probability):
  predict= {}
  for cls, cls_probability in prior_probability.items():
   predict[cls]= cls_probability
   for word, word_value in new_data.items():
     mean= stats[word][cls]["mean"]
     variance= stats[word][cls]["variance"]
     predict[cls]*= calculate_probability(word_value, mean, variance)

  ans= {}

  for final_key, final_value in predict.items():
    print(f"current word, predicted prob is: {final_key}, {final_value}")
    if len(ans) == 0:
      ans["Final ans"]= {
          "Key": final_key,
          "Prob": final_value
      }
    else:
      if final_value>ans["Final ans"]["Prob"]:
        ans["Final ans"]= {
          "Key": final_key,
          "Prob": final_value
        }

  return ans

```

```python
ans= predict(new_data, stats, calculate_probability, prior_probability)
print(ans)
```

```python
def evaluate(data, stats, calculate_probability, prior_probability):
  count= 0
  for row in range(data.shape[0]):
    new_data= {}
    for feature in data.columns:
      if feature == "Class":
        continue
      new_data[feature]= data[feature][row]
    current_prediction= predict(new_data, stats, calculate_probability, prior_probability)
    ans = current_prediction["Final ans"]["Key"]
    actual_ans= data["Class"][row]
    if ans== actual_ans:
      count+=1
  return count/data.shape[0]

```

```python
accuracy= evaluate(data, stats, calculate_probability, prior_probability)
print(accuracy)
```
